### Step 1: Import Library

```python
import pandas as pd
```

### Step 2: Load Dataset

```python
df = pd.read_csv("Sample - Superstore.csv", encoding="latin1")
```

### Step 3: Basic Information

```python
df.head()
df.info()
df.describe()
```

### Step 4: Finding Missing Values

```python
df.isnull().sum()
```

### Step 5: Remove Duplicates

```python
df = df.drop_duplicates()
```

### Step 6: Convert Date Columns

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

### Step 7: Save Cleaned Dataset

```python
df.to_csv("cleaned_superstore.csv", index=False)
```

### Step 8: Question 1 – Highest Sales Category

```python
category_sales = df.groupby("Category")["Sales"].sum()
print("\nTotal Sales by Category")
print(category_sales)
```

### Step 9: Question 2 – Highest Sales and Profit by Region

```python
region_analysis = df.groupby("Region")[["Sales", "Profit"]].sum()
print("\nSales and Profit by Region")
print(region_analysis)
```

### Step 10: Question 3 – Discount vs Profit

```python
discount_profit = df.groupby("Discount")[["Profit"]].mean()
print("\nAverage Profit by Discount")
print(discount_profit)
```
