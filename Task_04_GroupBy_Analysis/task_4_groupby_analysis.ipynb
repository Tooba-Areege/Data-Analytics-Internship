### Step 1: Import Pandas

```python
import pandas as pd
```

### Step 2: Load the cleaned dataset

```python
df= pd.read_csv("cleaned_titanic.csv")
```

### Step 3: Create the GroupBy Summary

```python
groupby_summary = df.groupby("Pclass").agg(
    Total_Passengers=("PassengerId", "count"),
    Average_Age=("Age", "mean"),
    Average_Fare=("Fare", "mean"),
    Survival_Rate=("Survived", "mean")
)
```

### Step 4: Convert survival rate into percentage

```python
groupby_summary["Survival_Rate"] = (
    groupby_summary["Survival_Rate"] * 100
).round(2)
```

```python
groupby_summary["Average_Age"] = (
    groupby_summary["Average_Age"].round(2)
)

groupby_summary["Average_Fare"] = (
    groupby_summary["Average_Fare"].round(2)
)
```

### Step 5: Display the result

```python
print("GroupBy Analysis: Passenger Class Comparison")
print(groupby_summary)
```

### Step 6: Save the summary

```python
groupby_summary.to_csv("passenger_class_analysis.csv")
```
