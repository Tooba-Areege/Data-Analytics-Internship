### Step 1: Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

### Step 2: Load Dataset

```python
df = pd.read_csv("cleaned_titanic.csv")
```

### Step 3: Create Charts

#### Chart 1: Suevival Count(Bar)

```python
plt.figure(figsize=(5,3))
df["Survived"].value_counts().plot(kind="bar")
plt.title("Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")
plt.tight_layout()
plt.savefig("survival_bar.png")
plt.show()
```

#### Chart 2: Average Age by Passenger Class (Line)

```python
plt.figure(figsize=(5,3))
df.groupby("Pclass")["Age"].mean().plot(kind="line", marker="o")
plt.title("Average Age by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Average Age")
plt.tight_layout()
plt.savefig("age_line.png")
plt.show()
```

#### Chart 3: Gender Distribution (Pie)

```python
plt.figure(figsize=(4,4))
df["Sex"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Gender Distribution")
plt.ylabel("")
plt.tight_layout()
plt.savefig("gender_pie.png")
plt.show()
```

#### Chart 4: Passenger Class Distribution (Bar)

```python
plt.figure(figsize=(5,3))
df["Pclass"].value_counts().sort_index().plot(kind="bar")
plt.title("Passenger Class Distribution")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.tight_layout()
plt.savefig("passenger_class_bar.png")
plt.show()
```
