 ### Step 1: IMPORT LIBRARIES

```python
import streamlit as st
import pandas as pd
import plotly.express as px
```

### Step 2: LOAD CLEANED DATASET


```python
df = pd.read_csv("cleaned_superstore.csv")
```

### Step 3: Convert date

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
```

### Step 4: Dashboard title

```python
st.title("📊 Superstore Sales & Profit Dashboard")

st.write(
    "Interactive analysis of sales, profit, product categories, "
    "regions, and discounts."
)
```

### Step 5:  Filters

```python
regions = st.multiselect(
    "Select Region",
    options=df["Region"].unique(),
    default=df["Region"].unique()
)

categories = st.multiselect(
    "Select Category",
    options=df["Category"].unique(),
    default=df["Category"].unique()
)

segments = st.multiselect(
    "Select Customer Segment",
    options=df["Segment"].unique(),
    default=df["Segment"].unique()
)
```

###  Step 6:  Filter dataset

```python
filtered_df = df[
    (df["Region"].isin(regions)) &
    (df["Category"].isin(categories)) &
    (df["Segment"].isin(segments))
]
```

### Step 7: Key metrics

```python
total_sales = filtered_df["Sales"].sum()
total_profit = filtered_df["Profit"].sum()
total_orders = filtered_df["Order ID"].nunique()
average_discount = filtered_df["Discount"].mean()
```

### Step 8: Display metrics

```python
col1, col2, col3, col4 = st.columns(4)

col1.metric("Total Sales", f"${total_sales:,.2f}")
col2.metric("Total Profit", f"${total_profit:,.2f}")
col3.metric("Total Orders", total_orders)
col4.metric("Average Discount", f"{average_discount:.2%}")
```

### Step 9 : Questions

 ### Question 1
 Which Product Category Generates the Highest Sales?"

```python
category_sales = (
    filtered_df
    .groupby("Category", as_index=False)["Sales"]
    .sum()
)

fig1 = px.bar(
    category_sales,
    x="Category",
    y="Sales",
    title="Sales by Product Category",
    text_auto=".2s"
)

st.plotly_chart(fig1, use_container_width=True)
```

###   Question 2
 Which Region Has the Highest Sales and Profit?


```python
region_analysis = (
    filtered_df
    .groupby("Region", as_index=False)[["Sales", "Profit"]]
    .sum()
)

fig2 = px.bar(
    region_analysis,
    x="Region",
    y=["Sales", "Profit"],
    barmode="group",
    title="Sales and Profit by Region"
)

st.plotly_chart(fig2, use_container_width=True)
```

###  Question 3: 
 How Do Discounts Affect Profit?


```python
discount_analysis = (
    filtered_df
    .groupby("Discount", as_index=False)["Profit"]
    .mean()
)

fig3 = px.line(
    discount_analysis,
    x="Discount",
    y="Profit",
    markers=True,
    title="Average Profit by Discount"
)

st.plotly_chart(fig3, use_container_width=True)
```

### Step 10 :Data table


```python
st.subheader("📋 Filtered Data")

st.dataframe(
    filtered_df,
    use_container_width=True
)
```
