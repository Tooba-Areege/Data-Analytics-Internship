### Step 1: Import Libraries 

```python
import pandas as pd
```

### Step 2: Load dataset

```python
df = pd.read_csv("cleaned_titanic.csv")
```

###  Step 3: Compare survival by gender

```python
comparison = df.groupby("Sex").agg(
    Total_Passengers=("PassengerId", "count"),
    Survival_Rate=("Survived", "mean")
)
```

### Step 4: Convert survival rate to percentage

```python
comparison["Survival_Rate"] = (
    comparison["Survival_Rate"] * 100
).round(2)
```

### Step 5: Display results

```python
print("Gender Survival Comparison")
print(comparison)
```

### step 6: Save results

```python
comparison.to_csv("gender_survival_comparison.csv")
```
