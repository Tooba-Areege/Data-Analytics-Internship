### Step 1: Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

### Step 2: Load Dataset

```python
df = pd.read_csv("cleaned_titanic.csv")
```

### Step 3: Calculate Key Metrics

```python
total_passengers = len(df)
survival_rate = round(df["Survived"].mean() * 100, 2)
average_age = round(df["Age"].mean(), 2)
average_fare = round(df["Fare"].mean(), 2)
```

### Step 4: Create Dashboard

```python
fig = plt.figure(figsize=(10,8))
```

### Step 5: Display Key Metrics

```python
plt.figtext(0.05,0.95,f"Total Passengers: {total_passengers}",fontsize=12)
plt.figtext(0.05,0.91,f"Survival Rate: {survival_rate}%",fontsize=12)
plt.figtext(0.05,0.87,f"Average Age: {average_age}",fontsize=12)
plt.figtext(0.05,0.83,f"Average Fare: {average_fare}",fontsize=12)
```

### Step 6: Add Survival Count

```python
ax1 = fig.add_axes([0.08,0.08,0.35,0.45])
df["Survived"].value_counts().plot(kind="bar", ax=ax1)
ax1.set_title("Survival Count")
ax1.set_xlabel("Survived")
ax1.set_ylabel("Passengers")
```

### Step 7: Add Passenger Class Chart

```python
ax2 = fig.add_axes([0.58,0.08,0.35,0.45])
df["Pclass"].value_counts().sort_index().plot(kind="bar", ax=ax2)
ax2.set_title("Passenger Class")
ax2.set_xlabel("Class")
ax2.set_ylabel("Passengers")
```

### Step 8: Save Dashboard

```python
plt.savefig("dashboard.png")
plt.show()

```
